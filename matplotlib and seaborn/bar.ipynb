```python
import numpy as np
import matplotlib.pyplot as plt

sachin = [460,972,204,513,315]
virat = [159,325,995,1381,1026]
virender = [893,810,446,645,217]
years = [2008,2009,2010,2011,2012]
x = np.arange(len(years))
width = 0.25
```

```python
plt.bar(x-width,sachin, width = width, label = 'sachin')
plt.bar(x, virat, width = width, label = 'virat')
plt.bar(x+width,virender, width= width, label = 'virender')
plt.xlabel("players")
plt.ylabel("Runs")
plt.title("Run Comparison")
plt.xticks(x,years)
plt.legend()
plt.tight_layout()
plt.show()
```
