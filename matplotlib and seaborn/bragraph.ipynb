```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
years = [1990, 1992, 1994, 1996, 1998, 2000, 2003, 2005, 2007, 2010]
runs =  [500, 700, 1100, 1500, 1800, 1200, 1700, 1300, 900, 1500]
```

```python
plt.style.available
```

```python
plt.bar(years , runs, color= "green",label = "Runs per yeas")
plt.plot(years , runs, 'ro--',label="Trend line")
plt.xlabel("years")
plt.ylabel("runs")
plt.legend() 
plt.title("sachine tendulakar years by run")
plt.style.use('seaborn-v0_8-bright',)
plt.grid(True)
plt.show()
```

```python
sachin = [500, 700, 1100, 1500, 1800, 1200, 1700, 1300, 900, 1500]
sehwag = [0, 200, 900, 1400, 1600, 1800, 1500, 1100, 800, 0]
kohli  = [0, 0, 500, 800, 1100, 1300, 1500, 1800, 1900, 2100]
```

```python
x = np.arange(len(years))
```

```python
width = 0.25
```

```python
plt.style.available
```

```python
plt.bar(x-width ,sachin, width=width,label = "sachin")
plt.plot(x-width,sachin,'bd--',label="sachin Trend line")
plt.bar(x,sehwag, width=width, label = "sehwag")
plt.plot(x,sehwag,'go--',label="sehwag Trend line")
plt.bar(x+width, kohli, width= width, label = "kohli")
plt.plot(x+width,kohli,'r^--',label="kohli Trend line")
plt.legend()
plt.xticks(x,years)
plt.style.use('seaborn-v0_8-deep',)
plt.xlabel("Years")
plt.ylabel("Runs")
#plt.grid(True)
plt.title("Runs compears")
```

```python
players = ["Sachin", "Sehwag", "Kohli"]
runs = [1500, 1200, 1800]
 
plt.bar(players, runs, color="skyblue")
plt.plot(players, runs, 'ro--')
# Add labels on top of bars
for i in range(len(players)):
    plt.text(i, runs[i] + 30, str(runs[i]), ha='center')
plt.grid(True)
```
