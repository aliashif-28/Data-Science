```python
years_col = [col for col in df.columns 
             if 'Year' in col][0]
kohli_col = [col for col in df.columns 
             if 'Kohli Runs' in col][0]
rohit_col = [col for col in df.columns 
             if 'Rohit Sharma Runs' in col][0]
sehwag_col = [col for col in df.columns 
              if 'Sehwag Runs' in col][0]
```

```python
years = df[years_col]
b = df[kohli_col]
c = df[rohit_col]
d = df[sehwag_col]
```

```python
plt.style.available
```

```python
plt.plot(b, 'ro--', label = 'Kohli')
plt.plot(c, 'g^--', label = 'Rohit')
plt.plot(d, 'bd--', label = 'Sehwag')
plt.xlabel("Years")
plt.ylabel("Runs")
plt.xticks(years)
plt.legend()
plt.title('Cricketer 10 Years Run Graph')
plt.tight_layout()
plt.style.use('seaborn-v0_8-muted',)

plt.show()
```
