```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
a = [20,15,10,30]
```

```python
b = [13,14,15,16]
```

```python
x = (a)
y = (b)
```

```python
plt.plot(x,'rd--', label = "ab")
plt.plot(y,'go--', label = "cd")
plt.xlabel('a,b')
plt.ylabel('c,d')
plt.title('Random graph')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
plt.plot(x,'rd--', label = "ab")
plt.plot(y,'go--', label = "cd")
plt.xlabel('a,b')
plt.ylabel('c,d')
plt.title('Random graph')
plt.legend()
plt.style.use('ggplot')
plt.tight_layout()
plt.show()
```

```python
with plt.xkcd():
    plt.plot(x,'rd--', label = "ab")
    plt.plot(y,'go--', label = "cd")
    plt.xlabel('a,b')
    plt.ylabel('c,d')
    plt.title('Random graph')
    plt.legend()
    #plt.style.use('seaborn-v0_8-bright')
    plt.grid(True)
    plt.tight_layout()
    plt.show()
```
