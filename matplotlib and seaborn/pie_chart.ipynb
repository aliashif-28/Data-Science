```python
print("Make the pie chart graph")
import matplotlib.pyplot as plt
labels = ['Samsung','i-Phone','Google pixal']
units = [1700000,920000,125000]
colors = ['#576C99','#E0A8ED','#DFF7BA']
explode = [0,0,0.1]
plt.pie(
    units, 
        labels = labels, 
        colors = colors, 
        explode = explode, 
        shadow = True, 
        autopct = '%1.1f%%'
       )
plt.title('selling phone in the last month for samsung, iphone, google pixal')
plt.show()
```

```python
import matplotlib.pyplot as plt
labels = ['Samsung','i-Phone','Google pixal']
units = [1700000,920000,125000]
for i in range(len(labels)):
    plt.text(i, units[i] + 80, str(units[i]), ha='center')
plt.bar(labels,units)
plt.xlabel("company name")
plt.ylabel("selling phone")
plt.title("selling phone in the last month for samsung, iphone, google pixal")
plt.show()
```

```python
import matplotlib.pyplot as plt
labels = ['Samsung','i-Phone','Google pixal']
units = [1700000,920000,125000]
plt.barh(labels,units)
plt.xlabel("company name")
plt.ylabel("selling phone")
plt.title("selling phone in the last month for samsung, iphone, google pixal")
plt.show()
```
