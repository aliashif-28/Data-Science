cheak the time python list vs numpy array

```python
import numpy as np
import time
```

Python List

```python
list1 = [1,2,3]
start = time.time()
list_squre = [x ** 2 for x in list1]
print(list_squre)
end = time.time()
print(end - start)
```

Numpy Arrar

```python
arr = np.array([1,2,3,])
start = time.time()
numpy_squre = arr ** 2
print(numpy_squre)
end = time.time()
print(end - start)
```
